```python
import pandas as pd
import seaborn as sns
from matplotlib import pyplot as plt

custom_params = {"axes.spines.right": False, "axes.spines.top": False}
sns.set_theme(context="paper", style="ticks", rc=custom_params, font="Arial", font_scale=1.5)
```

```python
final_protein_importances = pd.read_csv(
    "sepsis_feature_importances.tsv",
    sep="\t",
)
```

```python
fig, ax = plt.subplots()

sns.barplot(
    data=final_protein_importances,
    y="feature",
    x="mean_importance",
    color="#A32F3C",
    ax=ax,
)

fig.set_size_inches((3, 2.5))

ax.set_ylabel("")
ax.set_xlabel("Mean Importance")

ax.set_yticklabels(
    [label.get_text().split("_")[0] for label in ax.get_yticklabels()]
)

sns.despine()
```

```python
fig.figure.savefig(
    "sepsis_feature_importances.pdf", dpi=300, bbox_inches="tight"
)
```
