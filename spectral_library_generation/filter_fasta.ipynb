```python
from alphabase.peptide.fragment import get_charged_frag_types
import pandas as pd

from Bio import SeqIO
```

```python
human_atlas_library = pd.read_csv(
    "/Users/aaron/Documents/projects/sadt/data/fasta/hatlas_predicted_im_rt.tsv",
    sep="\t"
)
```

```python
hatlas_proteins = set(human_atlas_library['ProteinId'].unique())
```

```python
len(hatlas_proteins)
```

```python
filtered_proteins = []
missing_proteins = []

for seq_record in SeqIO.parse("/Users/aaron/Documents/projects/sadt/data/fasta/uniprotkb_taxonomy_id_9606_AND_reviewed_2025_08_21.fasta", "fasta"):
    
    if seq_record.id.split("|")[1] in hatlas_proteins:

        filtered_proteins.append(seq_record)

    else:

        missing_proteins.append(seq_record)
```

```python
len(filtered_proteins)
```

```python
filtered_protein_ids = set([protein.id.split("|")[1] for protein in filtered_proteins])
```

```python
hatlas_proteins - filtered_protein_ids
```

```python
SeqIO.write(filtered_proteins, "hatlas_filtered_fasta_250821.fasta", "fasta")
```
