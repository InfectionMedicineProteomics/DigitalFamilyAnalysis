```python
import re

def id_mod_sites(modified_peptide_sequence):

    tokens = re.split(r"\(UniMod:[0-9]+\)", modified_peptide_sequence)
    mod_site = 0
    mod_sites = []
    for token in tokens[:-1]:

        mod_site += len(token)

        mod_sites.append(str(mod_site))

    if len(mod_sites) > 1:

        return ";".join(mod_sites)
    
    elif len(mod_sites) == 1:

        return mod_sites[0]
    
    else:

        return ""
```

```python
precursors['mod_sites'] = precursors['ModifiedPeptideSequence'].apply(
    id_mod_sites
)
```

```python
precursors['mod_sites'] = precursors['mod_sites'].replace(-1, 'NaN')
precursors
```

```python
precursors['nAA'] = precursors['PeptideSequence'].str.len()
```

```python
extracted_mods = precursors['ModifiedPeptideSequence'].str.extractall(r'\((.*?)\)')
extracted_mods
```

```python
extracted_mods = extracted_mods.reset_index()

extracted_mods_idx = dict()

for idx, row in extracted_mods.iterrows():

    mods_ = extracted_mods_idx.get(row['level_0'], [])

    mods_.append(row[0])

    extracted_mods_idx[row['level_0']] = mods_

for key, value in extracted_mods_idx.items():

    if len(value) > 1:

        extracted_mods_idx[key] = ";".join(value)

    elif len(value) == 1:

        extracted_mods_idx[key] = value[0]

    else:

        extracted_mods_idx[key] = ""
```

```python
mods = pd.DataFrame(
    {
        "mods": extracted_mods_idx.values()
    },
    index=extracted_mods_idx.keys()
)
```

```python
precursors = precursors.join(
    mods
)
```

```python
precursors['precursor_id'] = precursors['ModifiedPeptideSequence'] + precursors['PrecursorCharge'].astype(str)
```

```python
predict_df = precursors[['precursor_id', 'PeptideSequence', 'PrecursorCharge', 'mods', 'mod_sites', 'nAA']].copy()
```

```python
predict_df = predict_df.rename(
    columns={
        "PeptideSequence": "sequence",
        "PrecursorCharge": "charge"
    }
)
predict_df
```

```python
predict_df = predict_df.fillna("")
predict_df
```
